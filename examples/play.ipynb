```python
import microegg as egg

E = egg.EGraph()

a = egg.Term.App("a", [])
b = egg.Term.App("b", [])
egg.Term.__add__ = lambda self, other: egg.Term.App("+", [self, other])

str(a + b)

# should probably not return a raw int
ab = E.add_term0(a+b)
ba = E.add_term0(b+a)
E.union(ab, ba)


E = egg.EGraph()
x,y = egg.Term.Var("x"), egg.Term.Var("y")
E.add_birule(x+y, y+x)
ab = E.add_term0(a+b)
E.run(1)
ba = E.add_term0(b+a)
E.extract(ab)
E.extract(ba)
```

```python
class MyType(egg.Term.App):
    def __init__(self, name):
        super().__init__(name, [])
    def __add__(self, other):
        return MyType("+", [self, other])
```
