```python
from microegg import EGraph, Term

E = EGraph()
a, b = Term.App("a", []), Term.App("b", [])
x, y = Term.Var("x"), Term.Var("y")
plus = lambda left, right: Term.App("+", [left, right])
E.add_rule(plus(x, y), plus(y, x))
ab = E.add_term(plus(a, b))
ba = E.add_term(plus(b, a))
E.run(2)
assert E.is_eq(ab, ba)
E.extract(ab)
```
